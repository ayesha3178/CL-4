```python
# MapReduce Program to Find Grades of Students
# Taking input from user

students = []

n = int(input("Enter number of students: "))

# INPUT PHASE
for i in range(n):
    name = input("\nEnter student name: ")
    marks = int(input("Enter marks: "))

    students.append((name, marks))

# MAP PHASE
mapped = []

for name, marks in students:

    if marks >= 90:
        grade = "A+"
    elif marks >= 80:
        grade = "A"
    elif marks >= 70:
        grade = "B"
    elif marks >= 60:
        grade = "C"
    elif marks >= 50:
        grade = "D"
    else:
        grade = "F"

    mapped.append((name, grade))

print("\nMapper Output:")
for item in mapped:
    print(item)

# REDUCE PHASE
reduced = {}

for name, grade in mapped:
    reduced[name] = grade

print("\nReducer Output:")
for name, grade in reduced.items():
    print(name, "->", grade)
```
