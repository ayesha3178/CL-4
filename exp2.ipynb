```python
# Input file reading

file = open("input.txt", "r")

text = file.read().lower().split()

# Given word
search_word = "hello"

# MAP PHASE
mapped = []

for word in text:
    if word == search_word:
        mapped.append((word, 1))

print("Mapper Output:")
print(mapped)

# REDUCE PHASE
count = 0

for key, value in mapped:
    count += value

print("\nReducer Output:")
print(search_word, "->", count)
```
