```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# -------------------------------
# EXTRACT
# -------------------------------

# Read Titanic dataset
data = pd.read_csv("titanic.csv")

print("Original Dataset:\n")
print(data.head())
```

```python
# -------------------------------
# TRANSFORM
# -------------------------------

# Remove missing values
data = data.dropna()

# Select required columns
data = data[['Sex', 'Survived']]

print("\nTransformed Dataset:\n")
print(data.head())
```

```python
# -------------------------------
# LOAD
# -------------------------------

# Save transformed data
data.to_csv("Processed_Titanic.csv", index=False)

print("\nData Loaded Successfully")
```

```python
# -------------------------------
# VISUALIZATION
# -------------------------------

# Count survival based on gender
survival_count = data.groupby('Sex')['Survived'].sum()

# Create bar chart
survival_count.plot(kind='bar')

plt.title("Survival Count by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Survivors")

plt.show()
```
