```python
import pandas as pd

df = pd.read_csv("titanic.csv")
df.head()
```

```python
from collections import defaultdict

def mapper(df):
    mapped = []

    for _, row in df.iterrows():
        sex = row['Sex']
        age = row['Age']
        survived = row['Survived']
        pclass = row['Pclass']

        # Problem 1: Male who died → emit age
        if sex == 'male' and survived == 0 and not pd.isna(age):
            mapped.append(("male_dead_age", age))

        # Problem 2: Female who died → emit class count
        if sex == 'female' and survived == 0:
            mapped.append((f"female_dead_class_{pclass}", 1))

    return mapped
```

```python
def shuffle(mapped):
    grouped = defaultdict(list)
    for key, value in mapped:
        grouped[key].append(value)
    return grouped
```

```python
def reducer(grouped):
    results = {}

    for key, values in grouped.items():
        # Average age for males who died
        if key == "male_dead_age":
            results["Average age of males who died"] = sum(values) / len(values)

        # Count females dead per class
        if key.startswith("female_dead_class"):
            results[key] = sum(values)

    return results
```

```python
mapped = mapper(df)
grouped = shuffle(mapped)
results = reducer(grouped)

results
```
