```python
import pyodbc
import pandas as pd

# Connect to SQL Server
conn = pyodbc.connect(
    'DRIVER={SQL Server};'
    'SERVER=localhost;'
    'DATABASE=TitanicDB;'
    'Trusted_Connection=yes;'
)

# SQL Query
query = "SELECT * FROM Titanic"

# Extract data
data = pd.read_sql(query, conn)

# Display data
print("Data Extracted from SQL Server:\n")
print(data.head())

# Load into CSV
data.to_csv("SQL_Titanic_Output.csv", index=False)

print("\nData loaded successfully")
```
